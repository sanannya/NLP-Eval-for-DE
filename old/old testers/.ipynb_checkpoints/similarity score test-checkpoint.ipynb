```python

import spacy
spacy.cli.download("en_core_web_lg")
```

```python
import spacy

nlp = spacy.load("en_core_web_lg")  # make sure to use larger package!
doc1 = nlp("the trash bag was ripped, so the user took the trash out and replaced the bag")
doc2 = nlp("the user dug through the unlabeled cabinets to find a smaller 12 V hand drill")

# Similarity of two documents
print(doc1.similarity(doc2))
```

```python
import spacy

nlp = spacy.load("en_core_web_lg")  # make sure to use larger package!
doc1 = nlp("Ripped trash bag")
doc2 = nlp("Unlabeled drill cabinet")

# Similarity of two documents
print(doc1.similarity(doc2))
```
