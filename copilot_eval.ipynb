```python
import csv
import pandas as pd
from NLP_Eval_for_DE import scores, data, results
```

```python
testable_data = data.get_testable_data("Example\\inputs\\case study 2 input-grouped by axials\\Hackathon features numerical.csv")
codes = data.get_codes("Example\\inputs\\case study 2 input-grouped by axials\\Hackathon features codes.csv")
copilotpredictions = pd.read_csv("hackathon_features_assigned_codes (1).csv")

#output the matrices
new_data = scores.evaluate(testable_data, copilotpredictions, codes)[0]
filename = "copilot_matrix" + '.csv'
with open(filename, 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(new_data)

#output the f1 scores

codes = codes["Codes"].tolist()
all_preds = []
all_preds.append(copilotpredictions)
f1s = []
for predictions in all_preds:
    f1s.append(scores.evaluate(testable_data, predictions, codes)[1])

new_data = []
codes.insert(0, "No code")
for i in range(len(codes)):
    new_data.append({"Code": codes[i], "Copilot F1": f1s[0][i]})

filename = "F1_scores" + '.csv'
fieldnames = ['Code', 'Copilot F1']

with open(filename, 'w', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(new_data)

#output the cohen's kappas
models = ["Copilot"]
kappas = []
for predictions in all_preds:
    kappas.append(scores.evaluate(testable_data, predictions, codes)[2])

new_data = []
for i in range(len(kappas)):
    new_data.append({"Model": models[i], "Cohen's Kappa": kappas[i]})

filename = "Cohens_kappas" + '.csv'
fieldnames = ['Model', "Cohen's Kappa"]

with open(filename, 'w', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(new_data)
```
